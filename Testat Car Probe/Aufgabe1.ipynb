```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator

# Vorgegebene Punkte aus der Skizze
x_punkte = np.array([1, 2, 3, 4, 5], dtype=float)
y_punkte = np.array([0, 0, 1, 0, 0], dtype=float)

# Vandermonde-Matrix: Spalten x^4, x^3, x^2, x und 1
A = np.vander(x_punkte, N=5)

# A @ koeffizienten = y_punkte
koeffizienten = np.linalg.solve(A, y_punkte)
a, b, c, d, e = koeffizienten

print("Koeffizienten [a, b, c, d, e]:")
print(np.round(koeffizienten, 6))
print(f"p(x) = {a:.2f}x⁴ {b:+.2f}x³ {c:+.2f}x² {d:+.2f}x {e:+.2f}")

# Prüfen, ob das Polynom tatsächlich durch die fünf Punkte geht
print("p(x) an den vorgegebenen Stellen:", np.round(np.polyval(koeffizienten, x_punkte), 6))
```

```python
# Viele x-Werte für eine glatte Kurve
x = np.linspace(0, 6.5, 800)
y = np.polyval(koeffizienten, x)

fig, ax = plt.subplots(figsize=(8, 4.8))

# Polynom und vorgegebene Punkte
ax.plot(x, y, color="blue", linewidth=1.6, label="Polynom")
ax.plot(x_punkte, y_punkte, "r*", markersize=10, label="Vorgegebene Punkte")

# Achsen, Gitterlinien und unverzerrte Darstellung
ax.set_xlim(0, 6.5)
ax.set_ylim(-1, 2.5)
ax.set_xticks(np.arange(0, 7, 1))
ax.set_yticks(np.arange(-1, 3, 1))
ax.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.yaxis.set_minor_locator(MultipleLocator(0.5))
ax.grid(True, which="major", alpha=0.35)
ax.grid(True, which="minor", alpha=0.15)
ax.set_aspect("equal", adjustable="box")

plt.show()
```
