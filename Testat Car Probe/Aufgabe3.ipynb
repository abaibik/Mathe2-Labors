```python
import numpy as np
import matplotlib.pyplot as plt

# Drehwinkel: 20 Grad gegen den Uhrzeigersinn
winkel = np.radians(20)

# Linienzüge des Buchstabens A
linien = [
    ([-3, 0, 3], [-3, 6, -3]),
    ([-2, 2], [0, 0])
]

fig, ax = plt.subplots()

for x, y in linien:
    x = np.array(x)
    y = np.array(y)

    # Ursprünglicher Buchstabe A
    ax.plot(x, y, color="black")

    # Rotation um den Ursprung
    x_rot = x * np.cos(winkel) - y * np.sin(winkel)
    y_rot = x * np.sin(winkel) + y * np.cos(winkel)

    ax.plot(x_rot, y_rot, color="red")

# Achsen und Gitternetz
ax.set_xlim(-4, 4.5)
ax.set_ylim(-5, 7.5)

ax.set_xticks(np.arange(-4, 5, 1))
ax.set_yticks(np.arange(-5, 8, 1))
ax.set_xticks(np.arange(-4, 4.5, 0.5), minor=True)
ax.set_yticks(np.arange(-5, 7.5, 0.5), minor=True)

ax.grid(True, which="both", alpha=0.3)
ax.set_aspect("equal")

plt.show()
```
