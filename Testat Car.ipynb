```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import AutoMinorLocator

# Gegebene Punkte
x_points = np.array([1, 2, 3])
y_points = np.array([1, 2, 2])

# Jede Zeile der Matrix hat die Form [x², x, 1]
A = np.column_stack((x_points**2, x_points, np.ones_like(x_points)))
a, b, c = np.linalg.solve(A, y_points)

print(f"a = {a:g}, b = {b:g}, c = {c:g}")

# Werte für die blaue Kurve berechnen
x = np.linspace(0, 4.5, 400)
y = a * x**2 + b * x + c

fig, ax = plt.subplots()
ax.plot(x, y, color="blue")
ax.plot(x_points, y_points, "r*", markersize=10)

ax.set_xlim(0, 4.5)
ax.set_ylim(0, 3.5)
ax.set_xticks(range(5))
ax.set_yticks(range(4))

ax.xaxis.set_minor_locator(AutoMinorLocator(2))
ax.yaxis.set_minor_locator(AutoMinorLocator(2))
ax.grid(which="major", alpha=0.5)
ax.grid(which="minor", alpha=0.2)
ax.set_axisbelow(True)
ax.set_aspect("equal")

plt.show()
```
